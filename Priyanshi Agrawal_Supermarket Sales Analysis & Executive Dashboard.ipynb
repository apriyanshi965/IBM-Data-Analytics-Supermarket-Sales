```python

import pandas as pd
import numpy as np 
import plotly.express as px

# Dataset read karein
df = pd.read_csv('SuperMarket Analysis.csv')

# Pehli 5 rows dekhne ke liye
df.head()
```

```python
# Data Summary Check
print("--- Missing Values Check ---")
print(df.isnull().sum())

print("\n--- Data Types Check ---")
print(df.dtypes)
```

```python
# Product Line wise Sales Revenue calculation
product_sales = df.groupby('Product line')['Total'].sum().reset_index()

# Plotly Bar Chart Banana
fig_bar = px.bar(
    product_sales, 
    x='Total', 
    y='Product line', 
    orientation='h', # Horizontal Bar Chart
    title='<b>Total Sales Revenue by Product Line</b>',
    color='Total',
    color_continuous_scale='Viridis'
)

# Graph display karna
fig_bar.show()
```

```python
# Exact Column Names dekhne ke liye
print(df.columns.tolist())
```

```python
# 1. Product line wise Total Sales calculate karein
product_sales = df.groupby('Product line')['Sales'].sum().reset_index()

# 2. Plotly Horizontal Bar Chart Banayein
fig_bar = px.bar(
    product_sales, 
    x='Sales', 
    y='Product line', 
    orientation='h', # Horizontal Bar
    title='<b>Total Sales Revenue by Product Line</b>',
    color='Sales',
    color_continuous_scale='Viridis',
    labels={'Sales': 'Total Revenue ($)', 'Product line': 'Product Category'}
)

# 3. Chart Display Karein
fig_bar.show()
```

```python
# Payment Method Share Pie Chart
fig_pie = px.pie(
    df, 
    names='Payment', 
    values='Sales', 
    title='<b>Sales Share by Payment Method</b>',
    hole=0.4, 
    color_discrete_sequence=px.colors.qualitative.Set3
)

fig_pie.show()
```

```python
# Key Business Metrics (KPIs) Calculation

total_revenue = df['Sales'].sum()
total_quantity = df['Quantity'].sum()
total_transactions = len(df)
avg_rating = df['Rating'].mean()


print(" SUPERMARKET EXECUTIVE KPIs SUMMARY")
print(f" Total Sales Revenue : ${total_revenue:,.2f}")
print(f" Total Items Sold     : {total_quantity:,} units")
print(f" Total Transactions   : {total_transactions:,}")
print(f" Average Customer Rating: {avg_rating:.2f} / 10")
```
